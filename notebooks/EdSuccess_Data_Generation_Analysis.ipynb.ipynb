## Attendance_Master_Data

```python
# ============================================================
# FINAL FACULTY SUBJECT COVERAGE + ATTENDANCE DATASET
# ============================================================

# 1. Get all unique subjects used across all programs
all_course_subjects = list(dict.fromkeys(
    subject
    for subject_list in courses_by_program.values()
    for subject in subject_list
))

print("Total unique subjects:", len(all_course_subjects))


# 2. Guarantee that every subject has at least one faculty
# First 25 faculty get one unique subject each

for i, subject in enumerate(all_course_subjects):
    faculty_df.loc[i, "Subject"] = subject


# Remaining faculty are distributed among valid course subjects
for i in range(len(all_course_subjects), len(faculty_df)):
    faculty_df.loc[i, "Subject"] = random.choice(all_course_subjects)


# 3. Validate Faculty coverage BEFORE Attendance generation

missing_subjects = (
    set(all_course_subjects)
    - set(faculty_df["Subject"])
)

print("Subjects without faculty:", missing_subjects)

if missing_subjects:
    raise ValueError(
        f"Faculty coverage incomplete: {missing_subjects}"
    )


# ============================================================
# 4. GENERATE SUBJECT-WISE ATTENDANCE
# ============================================================

attendance = []

for _, student in student_df.iterrows():

    student_id = student["Student_ID"]
    program = student["Program"]
    semester = student["Semester"]

    # Six valid subjects for the student's program
    valid_subjects = courses_by_program[program]

    for subject in valid_subjects:

        # Select only faculty teaching this exact subject
        faculty_options = faculty_df[
            faculty_df["Subject"] == subject
        ]

        faculty = faculty_options.sample(1).iloc[0]

        faculty_id = faculty["faculty_Id"]

        # Generate attendance figures
        classes_conducted = random.randint(30, 50)

        classes_attended = random.randint(
            15,
            classes_conducted
        )

        attendance_percentage = round(
            (classes_attended / classes_conducted) * 100,
            1
        )

        # Attendance status + risk
        if attendance_percentage >= 75:
            attendance_status = "Good"
            attendance_risk = "Low"

        elif attendance_percentage >= 60:
            attendance_status = "Average"
            attendance_risk = "Medium"

        else:
            attendance_status = "Poor"
            attendance_risk = "High"

        # Create attendance record
        attendance_record = {
            "Attendance_Id": f"ATT{len(attendance)+1:05}",
            "Student_ID": student_id,
            "Program": program,
            "Semester": semester,
            "Subject": subject,
            "Faculty_Id": faculty_id,
            "Classes_Conducted": classes_conducted,
            "Classes_Attended": classes_attended,
            "Attendance_Percentage": attendance_percentage,
            "Attendance_Status": attendance_status,
            "Attendance_Risk": attendance_risk
        }

        attendance.append(attendance_record)


# 5. Convert to DataFrame
attendance_df = pd.DataFrame(attendance)


# ============================================================
# 6. FINAL VALIDATION
# ============================================================

print("\n--- ATTENDANCE VALIDATION ---")

print("Attendance Shape:",
      attendance_df.shape)

print("Missing Faculty IDs:",
      attendance_df["Faculty_Id"].isna().sum())

print("Invalid Student IDs:",
      (~attendance_df["Student_ID"]
       .isin(student_df["Student_ID"])).sum())

print("Unique Students:",
      attendance_df["Student_ID"].nunique())

print("\nUnique subjects per student:")

print(
    attendance_df
    .groupby("Student_ID")["Subject"]
    .nunique()
    .value_counts()
)
```

## Issue_Tracker

```python
##Issue_Tracker
issue_categories = [
    "LMS Issue",
    "Fee Payment",
    "Academic Queries",
    "Assignment Issues",
    "Examination Queries",
    "Attendance Issue",
    "Login Problem",
    "Course Access ",
    "Certificate Request"
    "Admission Queries"
]

priorities = [
    "Low",
    "Medium",
    "High",
    "Critical"
]

ticket_statuses = [
    "Open",
    "In-Progress",
    "Resolved",
    "Closed",
    "Re-Opened"
]

assigned_team = [
    "Academic Team",
    "Student Service Team",
    "Finance Team",
    "Examination Team",
    "Sales Team",
    "Technical Team"
]

sla_days = {
    "Low": 5,
    "Medium": 3,
    "High": 2,
    "Critical": 1
}
```

```python
num_issues = 3000
issues = []
```

```python
for i in range(1,num_issues + 1):
    student = student_df.sample(1).iloc[0]
    program = student["Program"]
    semester = student["Semester"]
    issue_category = random.choice(issue_categories)
    priority = random.choice(priorities)
    ticket_status = random.choice(ticket_statuses)
    issue_description = "General student query"

    if issue_category in ["LMS Issue", "Login Problems", "Course Access"]:
        assigned_team = "Technical Support"
        
    elif issue_category in ["Fee Payment"]:
        assigned_team = "Finance Team"
        
    elif issue_category in ["Examination Queries"]:
        assigned_team = "Examination Team"
        
    elif issue_category in ["Academic Queries", "Assignment Issues"]:
        assigned_team = "Academic Team"
        
    elif issue_category in ["Certificate Request","Attendance Issue"]:
        assigned_team = "Student Service Team"
        
    elif issue_category == "Admission Queries":
        assigned_team = "Sales Team"
        
    issue_date = fake.date_between(start_date = "-1y",end_date= "today")
    
    if issue_category == "LMS Issue":
        issue_description = "Unable to access LMS"
        
    elif issue_category == "Fee Payment":
        issue_description = "Payment not reflecting in the student portal or LMS"
        
    elif issue_category == "Academic Queries":
        issue_description = "Student has an query related to course subject"
        
    elif issue_category == "Assignment Issues":
        issue_description = "Unable to submit assignment"

    elif issue_category == "Examination Queries":
        issue_description = "Student has a query regarding examination"

    elif issue_category == "Attendance Issue":
        issue_description = "Attendance record discrepancy"

    elif issue_category == "Login Problems":
        issue_description = "Unable to login to student portal"

    elif issue_category == "Course Access":
        issue_description = "Unable to access course content"

    elif issue_category == "Certificate Request":
        issue_description = "Student requested course certificate"

    elif issue_category == "Admission Queries":
        issue_description = "Student has an admission-related query"

    if ticket_status in ["Resolved", "Closed"]:
       resolution_tat = random.randint(1, 7)

       resolution_date = issue_date + pd.Timedelta(days=resolution_tat)
    
    else:
        resolution_tat = None
        resolution_date = None

    if resolution_tat is None:
       sla_status = "Pending"

    elif resolution_tat <= sla_days[priority]:
         sla_status = "Within SLA"

    else:
         sla_status = "SLA Breached"

    issue_record = {
        "Issue_Id" : f"ISS{i:04}",
        "Student_ID": student["Student_ID"],
        "Program": program,
        "Semester": semester,
        "Issue_Date": issue_date,
        "Issue_Category": issue_category,
        "Issue_Description": issue_description,
        "Priority": priority,
        "Assigned_Team": assigned_team,
        "Ticket_Status": ticket_status,
        "Resolution_Date": resolution_date,
        "Resolution_TAT_Days": resolution_tat,
        "SLA_Status": sla_status
    }
    issues.append(issue_record)
issue_df = pd.DataFrame(issues)
```

## Exam_Master_Data

```python
exam_results = []
```

```python
# FINAL EXAM RESULT DATASET

exam_results = []

for i, student in student_df.iterrows():

    student_id = student["Student_ID"]
    program = student["Program"]
    semester = student["Semester"]

    valid_subjects = courses_by_program[program]

    # Give each student an overall academic performance level
    performance_level = random.choices(
        ["High", "Average", "Low"],
        weights=[20, 50, 30],
        k=1
    )[0]

    for subject in valid_subjects:

        # Generate marks based on student's performance level
        if performance_level == "High":
            marks = random.randint(65, 100)

        elif performance_level == "Average":
            marks = random.randint(45, 80)

        else:
            marks = random.randint(25, 60)

        # Grade logic
        if marks >= 90:
            grade = "A+"
        elif marks >= 80:
            grade = "A"
        elif marks >= 70:
            grade = "B+"
        elif marks >= 60:
            grade = "B"
        elif marks >= 50:
            grade = "C+"
        elif marks >= 40:
            grade = "C"
        else:
            grade = "F"

        # Subject-level result
        if marks >= 40:
            result_status = "Pass"
        else:
            result_status = "Fail"

        exam_record = {
            "Exam_Result_ID": f"EXM{len(exam_results)+1:05}",
            "Student_ID": student_id,
            "Program": program,
            "Semester": semester,
            "Subject": subject,
            "Marks": marks,
            "Grade": grade,
            "Result_Status": result_status
        }

        exam_results.append(exam_record)

exam_result_df = pd.DataFrame(exam_results)
```

## Final_Results_Data

```python
# FINAL RESULT DATASET

final_results = []

for student_id, student_data in exam_result_df.groupby("Student_ID"):

    program = student_data["Program"].iloc[0]
    semester = student_data["Semester"].iloc[0]

    # Calculate overall marks
    total_marks = student_data["Marks"].sum()
    average_marks = round(student_data["Marks"].mean(), 1)

    # Student passes only if all subjects are passed
    if "Fail" in student_data["Result_Status"].values:
        final_result = "Fail"
    else:
        final_result = "Pass"

    # Final grade based on average marks
    if average_marks >= 90:
        final_grade = "A+"
    elif average_marks >= 80:
        final_grade = "A"
    elif average_marks >= 70:
        final_grade = "B+"
    elif average_marks >= 60:
        final_grade = "B"
    elif average_marks >= 50:
        final_grade = "C+"
    elif average_marks >= 40:
        final_grade = "D"
    else:
        final_grade = "F"

    final_record = {
        "Student_ID": student_id,
        "Program": program,
        "Semester": semester,
        "Total_Marks": total_marks,
        "Average_Marks": average_marks,
        "Final_Grade": final_grade,
        "Final_Result": final_result
    }

    final_results.append(final_record)

student_final_result_df = pd.DataFrame(final_results)
```
